## 2. Basic Matplotlib Vocabulary

Matplotlib has two common styles of writing code.

### The simple pyplot style

This is easy for beginners:

```python
plt.plot(x, y)
plt.show()
```

### The object-oriented style

This is more flexible and is commonly used in professional notebooks:

```python
fig, ax = plt.subplots()
ax.plot(x, y)
plt.show()
```

In this notebook, we will start simple and then use the object-oriented style for cleaner charts.

## 3. Plotting Data from Python Lists

A Python list is one of the simplest data sources for a plot.

````python
days = [1, 2, 3, 4, 5, 6, 7]
study_hours = [2, 3, 2.5, 4, 3.5, 5, 4.5]

plt.plot(days, study_hours)
plt.title("Study Hours Over One Week")
plt.xlabel("Day")
plt.ylabel("Study Hours")
plt.show()
````

### Adding markers and a grid

Markers help students see the actual data points.

````python
plt.plot(days, study_hours, marker="o")
plt.title("Study Hours Over One Week")
plt.xlabel("Day")
plt.ylabel("Study Hours")
plt.grid(True)
plt.show()
````

## 4. Plotting Data from NumPy Arrays

NumPy arrays are very common in data science and machine learning.
They are useful when we want to generate numerical sequences, simulations, or mathematical functions.

````python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title("Sine Function")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid(True)
plt.show()
````

### Multiple lines on the same chart

This is useful for comparing patterns.

````python
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label="sin(x)")
plt.plot(x, y2, label="cos(x)")
plt.title("Sine and Cosine")
plt.xlabel("x")
plt.ylabel("value")
plt.legend()
plt.grid(True)
plt.show()
````

## 5. The Object-Oriented Matplotlib Style

The object-oriented style gives us more control.

- `fig` represents the whole figure.
- `ax` represents the plotting area.

````python
fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(days, study_hours, marker="o")
ax.set_title("Study Hours Over One Week")
ax.set_xlabel("Day")
ax.set_ylabel("Study Hours")
ax.grid(True)

plt.show()
````

## 7. Line Plot

A line plot is useful when the x-axis has a natural order, such as time, days, or steps.

In data science, line plots are often used for:

- time series
- trends
- model training curves
- changes over ordered values

````python
df.groupby("exam_date")["final_score"].mean().hist()
````

````python
daily_average = (
    df.groupby("exam_date")["final_score"]
    .mean()
    .sort_index()
)

fig, ax = plt.subplots(figsize=(15, 5))

ax.plot(daily_average.index, daily_average.values, marker="o")
ax.set_title("Average Final Score by Exam Date")
ax.set_xlabel("Exam Date")
ax.set_ylabel("Average Final Score")
ax.grid(True)

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
````

## 8. Scatter Plot

A scatter plot is used to examine the relationship between two numerical variables.

In data science, scatter plots are useful for:

- checking relationships
- detecting clusters
- finding outliers
- exploring possible correlations

````python
fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(df["hours_studied"], df["final_score"], alpha=0.7)
ax.set_title("Hours Studied vs Final Score")
ax.set_xlabel("Hours Studied")
ax.set_ylabel("Final Score")
ax.grid(True)

plt.show()
````

### Scatter plot with a third variable

We can use point size to represent another numerical variable.

````python
fig, ax = plt.subplots(figsize=(8, 5))

point_sizes = df["attendance_rate"] * 2

ax.scatter(
    df["hours_studied"],
    df["final_score"],
    s=point_sizes,
    alpha=0.5
)

ax.set_title("Hours Studied vs Final Score")
ax.set_xlabel("Hours Studied")
ax.set_ylabel("Final Score")
ax.grid(True)

plt.show()
````

## 9. Bar Chart

A bar chart is useful for comparing categories.

In data science, bar charts are often used after `groupby`.

````python
average_by_major = (
    df.groupby("major")["final_score"]
    .mean()
    .sort_values(ascending=False)
)
````

````python
average_by_major.index
````

````python
average_by_major.values
````

````python
average_by_major = (
    df.groupby("major")["final_score"]
    .mean()
    .sort_values(ascending=False)
)

fig, ax = plt.subplots(figsize=(9, 5))

ax.bar(average_by_major.index, average_by_major.values)
ax.set_title("Average Final Score by Major")
ax.set_xlabel("Major")
ax.set_ylabel("Average Final Score")

plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
````

### Horizontal bar chart

Horizontal bar charts are useful when category names are long.

````python
fig, ax = plt.subplots(figsize=(8, 5))

ax.barh(average_by_major.index, average_by_major.values)
ax.set_title("Average Final Score by Major")
ax.set_xlabel("Average Final Score")
ax.set_ylabel("Major")

plt.tight_layout()
plt.show()
````

## 10. Histogram

A histogram shows the distribution of a numerical variable.

In data science, histograms help answer questions like:

- Is the data symmetric or skewed?
- Are there unusual values?
- Where are most values concentrated?

````python
fig, ax = plt.subplots(figsize=(8, 5))

ax.hist(df["final_score"], bins=12)
ax.set_title("Distribution of Final Scores")
ax.set_xlabel("Final Score")
ax.set_ylabel("Number of Students")

plt.show()
````

### Comparing distributions with multiple histograms

This can be useful, but students should be careful because overlapping histograms can become hard to read.

````python
fig, ax = plt.subplots(figsize=(8, 5))

ax.hist(df["previous_score"], bins=12, alpha=0.6, label="Previous Score")
ax.hist(df["final_score"], bins=12, alpha=0.6, label="Final Score")

ax.set_title("Previous Score vs Final Score Distribution")
ax.set_xlabel("Score")
ax.set_ylabel("Number of Students")
ax.legend()

plt.show()
````

## 11. Box Plot

A box plot summarizes the distribution of numerical data.

It is useful for showing:

- median
- spread
- possible outliers
- differences between groups

````python
fig, ax = plt.subplots(figsize=(7, 5))

ax.boxplot(df["final_score"], vert=True)
ax.set_title("Box Plot of Final Scores")
ax.set_ylabel("Final Score")
ax.set_xticks([1])
ax.set_xticklabels(["Students"])

plt.show()
````

### Box plot by category

This is useful for comparing the distribution of scores across groups.

````python
for mode, group_df in df.groupby("education_mode"):
    print(f"{mode}: {group_df['final_score'].mean()}")
````

````python
groups = []
labels = []

for mode, group_df in df.groupby("education_mode"):
    groups.append(group_df["final_score"])
    labels.append(mode)

fig, ax = plt.subplots(figsize=(8, 5))

ax.boxplot(groups, labels=labels)
ax.set_title("Final Score Distribution by Education Mode")
ax.set_xlabel("Education Mode")
ax.set_ylabel("Final Score")

plt.show()
````

## 12. Pie Chart

Pie charts are common, but they should be used carefully.

They are usually acceptable when:

- there are only a few categories
- the goal is to show parts of a whole
- exact comparison is not very important

For many data science tasks, a bar chart is often better.

````python
pass_counts = df["passed"].value_counts()

fig, ax = plt.subplots(figsize=(6, 6))

ax.pie(pass_counts.values, labels=pass_counts.index, autopct="%1.1f%%")
ax.set_title("Pass/Fail Percentage")

plt.show()
````

## 13. Grouped Bar Chart

Grouped bar charts are useful when comparing one numerical summary across two categorical variables.

````python
grouped = (
    df.groupby(["major", "education_mode"])["final_score"]
    .mean()
    .unstack()
)

fig, ax = plt.subplots(figsize=(10, 5))

x = np.arange(len(grouped.index))
width = 0.25

for i, column in enumerate(grouped.columns):
    ax.bar(x + i * width, grouped[column], width, label=column)

ax.set_title("Average Final Score by Major and Education Mode")
ax.set_xlabel("Major")
ax.set_ylabel("Average Final Score")
ax.set_xticks(x + width)
ax.set_xticklabels(grouped.index, rotation=30, ha="right")
ax.legend(title="Education Mode")

plt.tight_layout()
plt.show()
````

## 14. Error Bars

Error bars help show uncertainty or variability.

Here we show the mean final score by major with standard deviation.

````python
summary_by_major = df.groupby("major")["final_score"].agg(["mean", "std"]).sort_values("mean", ascending=False)

fig, ax = plt.subplots(figsize=(9, 5))

ax.bar(summary_by_major.index, summary_by_major["mean"], yerr=summary_by_major["std"], capsize=5)
ax.set_title("Mean Final Score by Major with Standard Deviation")
ax.set_xlabel("Major")
ax.set_ylabel("Final Score")

plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
````

## 15. Correlation Heatmap with Matplotlib

A correlation matrix helps us examine relationships between numerical variables.

A heatmap is a common way to visualize this matrix.

````python
numeric_columns = [
    "hours_studied",
    "attendance_rate",
    "sleep_hours",
    "assignments_completed",
    "previous_score",
    "final_score"
]

corr = df[numeric_columns].corr()

fig, ax = plt.subplots(figsize=(8, 6))

image = ax.imshow(corr)
fig.colorbar(image, ax=ax)

ax.set_title("Correlation Heatmap")
ax.set_xticks(np.arange(len(numeric_columns)))
ax.set_yticks(np.arange(len(numeric_columns)))
ax.set_xticklabels(numeric_columns, rotation=45, ha="right")
ax.set_yticklabels(numeric_columns)

for i in range(len(numeric_columns)):
    for j in range(len(numeric_columns)):
        ax.text(j, i, f"{corr.iloc[i, j]:.2f}", ha="center", va="center")

plt.tight_layout()
plt.show()
````

## 16. Subplots

Subplots let us show multiple charts in one figure.

This is useful for dashboards and quick exploratory analysis.

````python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

axes[0, 0].hist(df["final_score"], bins=12)
axes[0, 0].set_title("Final Score Distribution")
axes[0, 0].set_xlabel("Final Score")
axes[0, 0].set_ylabel("Count")

axes[0, 1].scatter(df["hours_studied"], df["final_score"], alpha=0.6)
axes[0, 1].set_title("Hours Studied vs Final Score")
axes[0, 1].set_xlabel("Hours Studied")
axes[0, 1].set_ylabel("Final Score")

major_means = df.groupby("major")["final_score"].mean().sort_values()
axes[1, 0].barh(major_means.index, major_means.values)
axes[1, 0].set_title("Average Score by Major")
axes[1, 0].set_xlabel("Average Final Score")

mode_counts = df["education_mode"].value_counts()
axes[1, 1].bar(mode_counts.index, mode_counts.values)
axes[1, 1].set_title("Students by Education Mode")
axes[1, 1].set_xlabel("Education Mode")
axes[1, 1].set_ylabel("Number of Students")

plt.tight_layout()
plt.show()
````

## 17. Working Directly with pandas Plotting

pandas has a convenient `.plot()` method that uses Matplotlib behind the scenes.

This is useful for quick exploration, but it is still important to understand Matplotlib because it gives more control.

````python
df["final_score"].plot(kind="hist", bins=12, figsize=(8, 5), title="Final Score Distribution from pandas")

plt.xlabel("Final Score")
plt.show()
````

````python
df.groupby("major")["final_score"].mean().sort_values().plot(
    kind="barh",
    figsize=(8, 5),
    title="Average Final Score by Major from pandas"
)

plt.xlabel("Average Final Score")
plt.ylabel("Major")
plt.show()
````

## 18. Saving a Plot

Plots can be saved as image files with `savefig`.

Common formats include:

- PNG

## 19. Choosing the Right Chart Type

| Question | Recommended Chart |
|---|---|
| How does a value change over time? | Line plot |
| What is the relationship between two numerical variables? | Scatter plot |
| How is one numerical variable distributed? | Histogram |
| How do categories compare? | Bar chart |
| Are there outliers or differences in spread? | Box plot |
| How are several numerical variables correlated? | Correlation heatmap |
| How can we compare categories across another category? | Grouped bar chart |
| What proportion does each category represent? | Pie chart, with caution |

## 20. Common Mistakes to Avoid

1. Forgetting `plt.show()` in scripts.
2. Using a line plot when the x-axis has unordered categories.
3. Using a pie chart with too many categories.
4. Not labeling axes.
5. Making charts too small.
6. Ignoring overlapping labels.
7. Showing too much information in one figure.
8. Choosing a chart before defining the question.
